```python
#list in python 
#List is an ordered and changeable collection that can store multiple values
marks = [80, 90, 75, 85]

print(marks)
```

```python
#acecessing elements in a list 
marks = [80, 90, 75, 85]

print(marks[0]) 
print(marks[1])
print(marks[2])
print(marks[3])
```

```python
#change elements in a list
marks = [80, 90, 75]

marks [1] = 95

print(marks)
```

```python
 #add elements to a list
marks = [80, 90, 75]

marks.append(85)

print(marks)
```

```python
#remove element from a list 
marks = [80, 90, 75 ]

marks.remove(90)

print(marks)
```

```python
#list of numbers 
numbers = [10, 20, 30]

numbers.insert(1, 15)

print(numbers)
```
