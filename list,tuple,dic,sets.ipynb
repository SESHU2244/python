```python
#tuples in python
#Tuple is a collection of multiple values that is ordered and cannot be changed after creation
student = ("seshadri", 98, "python")

print(student[0]) 
```

```python
#access value in a tuple 
student = ("seshadri", 21, 80.5)
print(student[0])
print(student[1])
```

```python
#immutable nature of tuples
student = ("seshadri", 21, 80.5)

print(1) = 21
#this gives an error because we cannot modify a tuple
```

```python
#tuple are immutable, meaning they cannot be changed after they are created. They are defined using parentheses ().

numbers = (10, 20, 20, 30, 20)

print(numbers.count(20))  
```

```python
numbers = (10, 20, 30, 40,)

print(numbers.index(30))
```

```python
numbers = (10, 20, 30, 40,)

print(len(numbers))
print(max(numbers))
print(min(numbers))
print(sum(numbers))
```

```python
#sets in python 
#Set is a collection unique values that is unordered and mutable
numbers = {10,20, 30, 20, 10}
print(numbers)
```

```python
#Why use sets 

#Suppose students have selected subjects
subjects = {"Python", "Java"}

subjects.add("SQL")

print(subjects)
```

```python
#remove values from a set 
subjects.remove("Java")

print(subjects)
```

```python
#Sets do not allow the duplicate values
numbers = {1, 2, 2, 3, 3, 4}

print(numbers)  
```

```python
#dictionaries in python
#Dictionary is a collection of key-value pairs that is unordered and mutable
student = {
    "name": "seshadri",
    "age":18,
    "course": "python"
}

print(student)
```

```python
#access elements in dic
print(student["name"])
print(student["age"])
print(student["course"])
```

```python
#add new data to a dictionary
student["city"] = "Tirupati"

print(student)
```
